```python
import json
from pathlib import Path
import pandas as pd


# Base paths 
from pathlib import Path

BASE_DIR = Path.cwd().parent  
DATA_DIR = BASE_DIR / "data"

csv_files = sorted(DATA_DIR.glob("nsf_grants_*.csv"))

print("Found CSV files:")
for f in csv_files:
    print("  -", f.name)
```

```python


# Find all yearly CSV files
csv_files = sorted(DATA_DIR.glob("nsf_grants_*.csv"))

merged_dfs = []  # Will store DataFrames for each year

for csv_path in csv_files:
    print(f"\nReading {csv_path.name} ...")

    # Extract year from filename, e.g. "nsf_grants_2021.csv" -> "2021"
    year = None
    try:
        year_str = csv_path.stem.split("_")[-1]
        if year_str.isnumeric():
            year = int(year_str)
    except Exception:
        pass

    if year is None:
        print(f"  Could not extract year from filename! Skipping {csv_path.name}")
        continue

    # Load one CSV
    df = pd.read_csv(csv_path)

    # Add year column
    df["year"] = year

    merged_dfs.append(df)

# Combine all DataFrames
print("\nMerging all years...")
df_all = pd.concat(merged_dfs, ignore_index=True)

# Save output
output_path = DATA_DIR / "nsf_grants_all_merged.csv"
df_all.to_csv(output_path, index=False)
```

```python
import altair as alt

# Selection that chooses the nearest year under the mouse
nearest = alt.selection_point(
    nearest=True,
    on='mouseover',
    fields=['year'],
    empty=False
)

# Base chart encoding (we'll reuse this in multiple layers)
base = alt.Chart(dir_year).encode(
    x=alt.X('year:O', title='Year'),
    y=alt.Y('n_grants:Q', title='# Grants'),
    color=alt.Color('org_dir_long_name:N', title='Directorate')
)

# 1) The main line for each directorate
line = base.mark_line().properties(
    width=800,   # wider plot like you asked
    height=400
)

# 2) Invisible selectors across the chart: capture the x (year) under the mouse
selectors = alt.Chart(dir_year).mark_point().encode(
    x='year:O',
    opacity=alt.value(0),
).add_params(
    nearest
)

# 3) Points on the lines, highlighted when nearest
points = base.mark_point().encode(
    opacity=alt.condition(nearest, alt.value(1), alt.value(0))
)

# 4) Text labels near the nearest points (show the total amount)
text = base.mark_text(align='left', dx=5, dy=-5).encode(
    text=alt.condition(nearest, 'n_grants:Q', alt.value(' '))
)

# 5) Vertical rule at the selected year
rules = alt.Chart(dir_year).mark_rule(color='gray').encode(
    x='year:O',
).transform_filter(
    nearest
)

chart_q2_line = alt.layer(
    line, selectors, points, rules, text
)

chart_q2_line
```

```python

# df_merged har: awd_id, year, dir_abbr, org_dir_long_name, awd_amount, status (0/1)

# Total grants per directorate & year (samme logik som din Q2)
dir_year_total = (
    df_merged
    .groupby(["dir_abbr", "org_dir_long_name", "year"], as_index=False)
    .agg(
        n_grants=("awd_id", "nunique"),      
        total_amount=("awd_amount", "sum")
    )
)

# Terminated grants per directorate & year
dir_year_term = (
    df_merged[df_merged["status"] == 1]
    .groupby(["dir_abbr", "org_dir_long_name", "year"], as_index=False)
    .agg(
        n_term=("awd_id", "nunique"),
        term_amount=("awd_amount", "sum")
    )
)

# Merge så vi har både total og terminated i samme tabel
dir_year_merged = dir_year_total.merge(
    dir_year_term,
    on=["dir_abbr", "org_dir_long_name", "year"],
    how="left"
)


# Label-tekst til overlay
dir_year_merged["term_label"] = (
    dir_year_merged["n_term"].fillna(0).astype(int).astype(str)
    + " terminated grants"
)
```

However, it quickly became apparent that the level of detail in the data was not enough and therefore for the secon iteration, the data was aggegated by month, allowing for a more detailed evolution througout the years. Furthermore, we split the moths and years up so it was possible to get a better overview. The y-axis containing the amount was also coverted into Milion USD for a cleaner look.

```python
import pandas as pd
import altair as alt

# -------------------------
# Prep + robust terminated flag
# -------------------------
df = df_merged.copy()
df["awd_eff_date"] = pd.to_datetime(df["awd_eff_date"], errors="coerce")
df = df.dropna(subset=["awd_eff_date", "state_x", "awd_id"])

df["month_date"] = df["awd_eff_date"].dt.to_period("M").dt.to_timestamp()
df["year"] = df["month_date"].dt.year.astype(int)

# Keep same range as Q4 (adjust if you want)
df = df[(df["year"] >= 2021) & (df["year"] <= 2025)]

s = df["status"]
df["is_term"] = (
    (s == True) |
    (s == 1) |
    (s.astype(str).str.strip().str.lower().isin(
        ["true", "t", "1", "yes", "y", "terminated", "cancelled", "canceled"]
    ))
)

# -------------------------
# 1) Termination states per month (for tooltip)
# -------------------------
df_term_states = (
    df[df["is_term"]]
    .groupby("month_date")["state_x"]
    .apply(lambda x: ", ".join(sorted(x.dropna().unique())))
    .reset_index(name="term_states")
)

df_term_states["n_term_states"] = df_term_states["term_states"].apply(
    lambda txt: 0 if (pd.isna(txt) or txt == "") else len(txt.split(", "))
)

# -------------------------
# 2) Monthly aggregation (COUNTS)
# -------------------------
tot = (df.groupby(["state_x", "month_date"], as_index=False)
         .agg(n_grants=("awd_id", "count")))

term = (df[df["is_term"]]
          .groupby(["state_x", "month_date"], as_index=False)
          .agg(n_term=("awd_id", "count")))

df_q5m = tot.merge(term, on=["state_x", "month_date"], how="left").fillna({"n_term": 0})
df_q5m["year"] = df_q5m["month_date"].dt.year.astype(int)
df_q5m["year_str"] = df_q5m["year"].astype(str)

# Complete months so timeline is continuous per state
states = sorted(df_q5m["state_x"].dropna().unique().tolist())
months = pd.date_range("2021-01-01", "2025-12-01", freq="MS")

full_grid = pd.MultiIndex.from_product([states, months], names=["state_x", "month_date"]).to_frame(index=False)
full_grid["year"] = full_grid["month_date"].dt.year.astype(int)
full_grid["year_str"] = full_grid["year"].astype(str)

df_q5m_full = (full_grid.merge(df_q5m, on=["state_x", "month_date", "year", "year_str"], how="left")
               .fillna({"n_grants": 0, "n_term": 0}))

# Merge in termination-state list (same for all states; used mainly when "All")
df_q5m_full = df_q5m_full.merge(df_term_states, on="month_date", how="left")
df_q5m_full["term_states"] = df_q5m_full["term_states"].fillna("")
df_q5m_full["n_term_states"] = df_q5m_full["n_term_states"].fillna(0).astype(int)

# -------------------------
# 3) Controls: dropdown + toggle + zoom
# -------------------------
state_param = alt.param(
    name="SelectedState",
    value="All",
    bind=alt.binding_select(options=["All"] + states, name="Select state: ")
)

show_term = alt.param(
    name="ShowTerminated",
    value=True,
    bind=alt.binding_checkbox(name="Show terminated bars ")
)

zoom = alt.selection_interval(bind="scales", encodings=["x"], name="Zoom")

# Filter by state (or All)
base = alt.Chart(df_q5m_full).add_params(state_param, show_term).transform_filter(
    "SelectedState == 'All' || datum.state_x == SelectedState"
)

# Aggregate to ONE value per month_date (so "All" becomes a single time series)
agg = base.transform_aggregate(
    n_grants="sum(n_grants)",
    n_term="sum(n_term)",
    n_term_states="max(n_term_states)",   # same for all states -> max is fine
    term_states="max(term_states)",       # same for all states -> max is fine
    groupby=["month_date"]
)

# Show termination states/counts only when "All" is selected
agg = agg.transform_calculate(
    term_states_show="SelectedState == 'All' ? datum.term_states : ''",
    n_term_states_show="SelectedState == 'All' ? datum.n_term_states : null"
)




# -------------------------
# 4) Two-level x-axis (years above, months below)
# Use a separate timeline to avoid duplicates when "All"
# -------------------------
timeline = pd.DataFrame({"month_date": months})
timeline["year_str"] = timeline["month_date"].dt.year.astype(int).astype(str)

year_labels = alt.Chart(timeline).transform_filter(
    "month(datum.month_date) == 0"
).mark_text(dy=-5).encode(
    x=alt.X("month_date:T", axis=None),
    text="year_str:N"
).properties(width=850, height=28)

year_lines = alt.Chart(timeline).transform_filter(
    "month(datum.month_date) == 0"
).mark_rule(color="lightgray").encode(
    x="month_date:T"
)

# -------------------------
# 5) Main layers: line + dots + terminated bars (dual y-axis)
# -------------------------

month_ticks = months.tolist()  # month starts from 2021-01-01 to 2025-12-01

line_total = agg.mark_line().encode(
    x=alt.X(
    "month_date:T",
    title="",
    axis=alt.Axis(
        values=month_ticks,     # <- only month-start ticks
        format="%b",            # Jan, Feb, ...
        labelAngle=0,
        grid=False
    )
),
    y=alt.Y("n_grants:Q", title="# grants", axis=alt.Axis(format=",.0f")),
    tooltip=[
    alt.Tooltip("month_date:T", title="Month", format="%Y-%m"),
    alt.Tooltip("n_grants:Q", title="# grants", format=",.0f"),
    alt.Tooltip("n_term:Q", title="# terminated (same month)", format=",.0f"),
    alt.Tooltip("n_term_states_show:Q", title="# states w/ terminations", format=",.0f"),
    alt.Tooltip("term_states_show:N", title="States w/ terminations"),
]

).add_selection(zoom).properties(width=850, height=300)

points = agg.mark_point(size=45).encode(
    x="month_date:T",
    y=alt.Y("n_grants:Q", axis=None)   # <- disable axis so it doesn't duplicate the left title
)


term_bars = agg.transform_filter(
    "datum.n_term > 0"
).mark_bar(color="#E45756").encode(
    x="month_date:T",
    y=alt.Y(
        "n_term:Q",
        title="# terminated grants",
        axis=alt.Axis(orient="right", format=",.0f", tickMinStep=1)
    ),
    opacity=alt.condition(show_term, alt.value(0.7), alt.value(0.0)),
    tooltip=[
        alt.Tooltip("month_date:T", title="Month", format="%Y-%m"),
        alt.Tooltip("n_term:Q", title="# terminated", format=",.0f"),
        alt.Tooltip("n_term_states_show:Q", title="# states w/ terminations", format=",.0f"),
        alt.Tooltip("term_states_show:N", title="States w/ terminations"),

    ]
)

main_panel = (year_lines + term_bars + line_total + points).resolve_scale(y="independent")

q5_final = alt.vconcat(
    year_labels,
    main_panel,
    spacing=0
).resolve_scale(x="shared").properties(
    title="Q5 — Evolution of the sum of grants for each state and terminated grants (bars) — drag to zoom"
)

q5_final
```
