```python
import json
from pathlib import Path
import pandas as pd


# Base paths 
from pathlib import Path

BASE_DIR = Path.cwd().parent  
DATA_DIR = BASE_DIR / "data"

csv_files = sorted(DATA_DIR.glob("nsf_grants_*.csv"))

print("Found CSV files:")
for f in csv_files:
    print("  -", f.name)
```

```python


# Find all yearly CSV files
csv_files = sorted(DATA_DIR.glob("nsf_grants_*.csv"))

merged_dfs = []  # Will store DataFrames for each year

for csv_path in csv_files:
    print(f"\nReading {csv_path.name} ...")

    # Extract year from filename, e.g. "nsf_grants_2021.csv" -> "2021"
    year = None
    try:
        year_str = csv_path.stem.split("_")[-1]
        if year_str.isnumeric():
            year = int(year_str)
    except Exception:
        pass

    if year is None:
        print(f"  Could not extract year from filename! Skipping {csv_path.name}")
        continue

    # Load one CSV
    df = pd.read_csv(csv_path)

    # Add year column
    df["year"] = year

    merged_dfs.append(df)

# Combine all DataFrames
print("\nMerging all years...")
df_all = pd.concat(merged_dfs, ignore_index=True)

# Save output
output_path = DATA_DIR / "nsf_grants_all_merged.csv"
df_all.to_csv(output_path, index=False)
```

```python
import altair as alt
from vega_datasets import data

# Base map of the US
map_data = alt.topo_feature(data.us_10m.url, feature="states")

usChart = alt.Chart(map_data).mark_geoshape(
    fill="lightgray",
    stroke="white"
).properties(
    width=600,
    height=360
).project("albersUsa")
```

```python
capitals = data.us_state_capitals.url

# Unique list of years for the dropdown
years = sorted(state_year["year"].dropna().unique().tolist())

# A dropdown selector for the year (this is like the interactions in your tutorial notebook)
year_dropdown = alt.binding_select(options=years, name="Year: ")

year_selection = alt.selection_point(
    fields=["year"],
    bind=year_dropdown
)
 # start with the latest year

points_base = alt.Chart(state_year, title="NSF Grants per State and Year"
).transform_lookup(
    lookup="state_name",
    from_=alt.LookupData(capitals, key="state", fields=["state", "lat", "lon"])
).transform_filter(
    alt.datum.lat != None
)

points = points_base.add_params(
    year_selection
).transform_filter(
    year_selection
).mark_circle(
    size=120,
    opacity=0.9,
    stroke="black",
    strokeWidth=0.6
).encode(
    longitude="lon:Q",
    latitude="lat:Q",
    color=alt.Color(
        "n_grants:Q",
        title="Number of grants",
        scale=alt.Scale(range=["#F1F4F7", "#B5D5DB", "#78B8C4", "#2F7F94"])
    ),
    tooltip=[
        alt.Tooltip("state_name:N", title="State"),
        alt.Tooltip("year:O", title="Year"),
        alt.Tooltip("n_grants:Q", title="# Grants"),
        alt.Tooltip("total_amount:Q", title="Total amount", format=",.0f")
    ]
)

chart_q1 = (usChart + points).configure_view(stroke=None)
chart_q1
```

```python
# 3) Data med statskapitaler (lat/lon)
capitals = data.us_state_capitals.url

# 4) Dropdown med år
years = sorted(state_year["year"].dropna().unique().tolist())
year_dropdown = alt.binding_select(options=years, name="Year: ")

year_selection = alt.selection_point(
    fields=["year"],
    bind=year_dropdown
)

# 5) Bobler uden baggrundskort – men stadig geografisk placeret
points_base = (
    alt.Chart(state_year, title="NSF Grants per State and Year")
    .transform_lookup(
        lookup="state_name",  # fuldt navn til lookup
        from_=alt.LookupData(capitals, key="state", fields=["state", "lat", "lon"])
    )
    .transform_filter(
        alt.datum.lat != None  # smid stater uden lat/lon væk
    )
    .properties(
        width=400,
        height=300
    )
    .project("albersUsa")      # samme proj som et US-map, bare uden geoshape
)

points = (
    points_base
    .add_params(year_selection)
    .transform_filter(year_selection)
    .mark_circle(
        opacity=1,
        stroke="black",
        strokeWidth=0.8
    )
    .encode(
        longitude="lon:Q",
        latitude="lat:Q",
        text="state_x:N",
        size=alt.Size(
            "n_grants:Q",
            title="# Grants",
            scale=alt.Scale(range=[50, 1000])
        ),
        color=alt.Color(
            "n_grants:Q",
            title="Number of grants",
            scale=alt.Scale(range=["#F1F4F7", "#B5D5DB", "#78B8C4", "#2F7F94"])
        ),
        tooltip=[
            alt.Tooltip("state_x:N", title="State"),  
            alt.Tooltip("state_name:N", title="State (full name)"),
            alt.Tooltip("year:O", title="Year"),
            alt.Tooltip("n_grants:Q", title="# Grants"),
            alt.Tooltip("total_amount:Q", title="Total amount", format=",.0f")
        ]
    ))

text_labels = (
    points_base
    .add_params(year_selection)
    .transform_filter(year_selection)
    .mark_text(
        fontSize=10,
        dy=-6,            
        color="black"
    )
    .encode(
        longitude="lon:Q",
        latitude="lat:Q",
        text="state_x:N"  
    )


)

chart_q1 = alt.layer(points, text_labels).configure_view(stroke=None)

chart_q1 
```

```python



dir_year = (
    df_merged
    .groupby(["dir_abbr", "org_dir_long_name", "year"], as_index=False)
    .agg(
        n_grants=("awd_id", "nunique"),
        total_amount=("awd_amount", "sum")
    )
)

# Unique list of years for dropdown
years = sorted(dir_year["year"].dropna().unique().tolist())

# Year selection (Altair 5 style)
year_dropdown = alt.binding_select(options=years, name="Year: ")
year_sel = alt.selection_point(fields=["year"], bind=year_dropdown)

chart_q2 = (
    alt.Chart(dir_year, title="NSF Grants per Directorate")
    .add_params(year_sel)
    .transform_filter(year_sel)
    .mark_bar()
    .encode(
        x=alt.X("n_grants:Q", title="# Grants"),
        y=alt.Y(
            "org_dir_long_name:N",
            sort="-x",
            title="Directorate"
        )
      
        ,
        tooltip=[
            alt.Tooltip("org_dir_long_name:N", title="Directorate"),
            alt.Tooltip("dir_abbr:N", title="Abbr"),
            alt.Tooltip("year:O", title="Year"),
            alt.Tooltip("n_grants:Q", title="# Grants"),
            alt.Tooltip("total_amount:Q", title="Total amount", format=",.0f"),
        ]
    )
    .properties(width=450, height=300)
)

chart_q2
```

```python
import altair as alt

# Selection that chooses the nearest year under the mouse
nearest = alt.selection_point(
    nearest=True,
    on='mouseover',
    fields=['year'],
    empty=False
)

# Base chart encoding (we'll reuse this in multiple layers)
base = alt.Chart(dir_year).encode(
    x=alt.X('year:O', title='Year'),
    y=alt.Y('n_grants:Q', title='# Grants'),
    color=alt.Color('org_dir_long_name:N', title='Directorate')
)

# 1) The main line for each directorate
line = base.mark_line().properties(
    width=800,   # wider plot like you asked
    height=400
)

# 2) Invisible selectors across the chart: capture the x (year) under the mouse
selectors = alt.Chart(dir_year).mark_point().encode(
    x='year:O',
    opacity=alt.value(0),
).add_params(
    nearest
)

# 3) Points on the lines, highlighted when nearest
points = base.mark_point().encode(
    opacity=alt.condition(nearest, alt.value(1), alt.value(0))
)

# 4) Text labels near the nearest points (show the total amount)
text = base.mark_text(align='left', dx=5, dy=-5).encode(
    text=alt.condition(nearest, 'n_grants:Q', alt.value(' '))
)

# 5) Vertical rule at the selected year
rules = alt.Chart(dir_year).mark_rule(color='gray').encode(
    x='year:O',
).transform_filter(
    nearest
)

chart_q2_line = alt.layer(
    line, selectors, points, rules, text
)

chart_q2_line
```

## Ex 2+3

```python

# df_merged har: awd_id, year, dir_abbr, org_dir_long_name, awd_amount, status (0/1)

# Total grants per directorate & year (samme logik som din Q2)
dir_year_total = (
    df_merged
    .groupby(["dir_abbr", "org_dir_long_name", "year"], as_index=False)
    .agg(
        n_grants=("awd_id", "nunique"),      
        total_amount=("awd_amount", "sum")
    )
)

# Terminated grants per directorate & year
dir_year_term = (
    df_merged[df_merged["status"] == 1]
    .groupby(["dir_abbr", "org_dir_long_name", "year"], as_index=False)
    .agg(
        n_term=("awd_id", "nunique"),
        term_amount=("awd_amount", "sum")
    )
)

# Merge så vi har både total og terminated i samme tabel
dir_year_merged = dir_year_total.merge(
    dir_year_term,
    on=["dir_abbr", "org_dir_long_name", "year"],
    how="left"
)


# Label-tekst til overlay
dir_year_merged["term_label"] = (
    dir_year_merged["n_term"].fillna(0).astype(int).astype(str)
    + " terminated grants"
)
```

```python

# Checkbox parameter to toggle terminated overlay
show_term = alt.param(
    value=False,
    bind=alt.binding_checkbox(name="Show terminated grants")
)
```

```python
base_bars = (
    alt.Chart(dir_year_merged, title="NSF Grants per Directorate")
    .add_params(year_sel, show_term)     # <-- year + checkbox i samme chart
    .transform_filter(year_sel)
    .mark_bar()
    .encode(
        x=alt.X("n_grants:Q", title="# Grants"),   # samme som din Q2
        y=alt.Y(
            "org_dir_long_name:N", sort="-x", title="Directorate"),
        color=alt.value("#4C78A8"),
        tooltip=[
            alt.Tooltip("org_dir_long_name:N", title="Directorate"),
            alt.Tooltip("dir_abbr:N", title="Abbr"),
            alt.Tooltip("year:O", title="Year"),
            alt.Tooltip("n_grants:Q", title="# Grants"),
            alt.Tooltip("total_amount:Q", title="Total amount", format=",.0f"),
        ]
    )
    .properties(width=450, height=300)
)



# Røde termination-bars – samme bredde som de blå
term_bars = (
    alt.Chart(dir_year_merged)
    .transform_filter(year_sel)
    .mark_bar()
    .encode(
        x=alt.X("n_term:Q", title="# Terminated Grants"),
        y=alt.Y("org_dir_long_name:N",sort="-x", title="Directorate"),
        color=alt.value("#E45756"),
        opacity=alt.condition(show_term, alt.value(0.8), alt.value(0.0)),
        tooltip=[
            alt.Tooltip("org_dir_long_name:N", title="Directorate"),
            alt.Tooltip("year:O", title="Year"),
            alt.Tooltip("n_term:Q", title="# Terminated Grants"),
            alt.Tooltip("term_amount:Q", title="Terminated amount", format=",.0f")
        ]
    )
)

# Tekst ved enden af søjlen: "27 terminated grants"
term_text = (
    alt.Chart(dir_year_merged)
    .transform_filter(year_sel)            
    .transform_filter("datum.n_term > 0")  # kun hvis der er noget at vise
    .mark_text(
        align="left",
        dx=5,
        baseline="middle"
    )
    .encode(
        x="n_grants:Q",                    # text start ved enden af blå søjle
        y=alt.Y("org_dir_long_name:N", sort="-x", title="Directorate"),
        text="term_label:N",
        opacity=alt.condition(
            show_term,
            alt.value(1),
            alt.value(0)
        )
    )
)
```

```python
chart_q2_q3 = (
    alt.layer(base_bars, term_bars, term_text)
)

chart_q2_q3
```
