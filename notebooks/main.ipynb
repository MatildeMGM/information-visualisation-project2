```python
import json
from pathlib import Path
import pandas as pd


# Base paths 
from pathlib import Path

BASE_DIR = Path.cwd().parent  
DATA_DIR = BASE_DIR / "data"

csv_files = sorted(DATA_DIR.glob("nsf_grants_*.csv"))

print("Found CSV files:")
for f in csv_files:
    print("  -", f.name)
```

```python


# Find all yearly CSV files
csv_files = sorted(DATA_DIR.glob("nsf_grants_*.csv"))

merged_dfs = []  # Will store DataFrames for each year

for csv_path in csv_files:
    print(f"\nReading {csv_path.name} ...")

    # Extract year from filename, e.g. "nsf_grants_2021.csv" -> "2021"
    year = None
    try:
        year_str = csv_path.stem.split("_")[-1]
        if year_str.isnumeric():
            year = int(year_str)
    except Exception:
        pass

    if year is None:
        print(f"  Could not extract year from filename! Skipping {csv_path.name}")
        continue

    # Load one CSV
    df = pd.read_csv(csv_path)

    # Add year column
    df["year"] = year

    merged_dfs.append(df)

# Combine all DataFrames
print("\nMerging all years...")
df_all = pd.concat(merged_dfs, ignore_index=True)

# Save output
output_path = DATA_DIR / "nsf_grants_all_merged.csv"
df_all.to_csv(output_path, index=False)
```

```python


state_year = (
    df_clean
    .groupby(["state", "year"], as_index=False)
    .agg(
        n_grants=("awd_id", "nunique"),
        total_amount=("awd_amount", "sum")
    )
)
# Map state abbreviation to full state name (needed for us_state_capitals lookup)
abbr2name = {
    'AL':'Alabama','AK':'Alaska','AZ':'Arizona','AR':'Arkansas','CA':'California','CO':'Colorado',
    'CT':'Connecticut','DE':'Delaware','FL':'Florida','GA':'Georgia','HI':'Hawaii','ID':'Idaho',
    'IL':'Illinois','IN':'Indiana','IA':'Iowa','KS':'Kansas','KY':'Kentucky','LA':'Louisiana',
    'ME':'Maine','MD':'Maryland','MA':'Massachusetts','MI':'Michigan','MN':'Minnesota','MS':'Mississippi',
    'MO':'Missouri','MT':'Montana','NE':'Nebraska','NV':'Nevada','NH':'New Hampshire','NJ':'New Jersey',
    'NM':'New Mexico','NY':'New York','NC':'North Carolina','ND':'North Dakota','OH':'Ohio','OK':'Oklahoma',
    'OR':'Oregon','PA':'Pennsylvania','RI':'Rhode Island','SC':'South Carolina','SD':'South Dakota',
    'TN':'Tennessee','TX':'Texas','UT':'Utah','VT':'Vermont','VA':'Virginia','WA':'Washington',
    'WV':'West Virginia','WI':'Wisconsin','WY':'Wyoming','DC':'District of Columbia'
}

state_year["state_name"] = state_year["state"].map(abbr2name)
```

```python
import altair as alt
from vega_datasets import data

# Base map of the US
map_data = alt.topo_feature(data.us_10m.url, feature="states")

usChart = alt.Chart(map_data).mark_geoshape(
    fill="lightgray",
    stroke="white"
).properties(
    width=600,
    height=360
).project("albersUsa")
```

```python
capitals = data.us_state_capitals.url

# Unique list of years for the dropdown
years = sorted(state_year["year"].dropna().unique().tolist())

# A dropdown selector for the year (this is like the interactions in your tutorial notebook)
year_dropdown = alt.binding_select(options=years, name="Year: ")

year_selection = alt.selection_point(
    fields=["year"],
    bind=year_dropdown
)
 # start with the latest year

points_base = alt.Chart(state_year, title="NSF Grants per State and Year"
).transform_lookup(
    lookup="state_name",
    from_=alt.LookupData(capitals, key="state", fields=["state", "lat", "lon"])
).transform_filter(
    alt.datum.lat != None
)

points = points_base.add_params(
    year_selection
).transform_filter(
    year_selection
).mark_circle(
    size=120,
    opacity=0.9,
    stroke="black",
    strokeWidth=0.6
).encode(
    longitude="lon:Q",
    latitude="lat:Q",
    color=alt.Color(
        "n_grants:Q",
        title="Number of grants",
        scale=alt.Scale(range=["#F1F4F7", "#B5D5DB", "#78B8C4", "#2F7F94"])
    ),
    tooltip=[
        alt.Tooltip("state_name:N", title="State"),
        alt.Tooltip("year:O", title="Year"),
        alt.Tooltip("n_grants:Q", title="# Grants"),
        alt.Tooltip("total_amount:Q", title="Total amount", format=",.0f")
    ]
)

chart_q1 = (usChart + points).configure_view(stroke=None)
chart_q1
```

```python
import altair as alt
from vega_datasets import data

capitals = data.us_state_capitals.url

# Unique list of years for the dropdown
years = sorted(state_year["year"].dropna().unique().tolist())

year_dropdown = alt.binding_select(options=years, name="Year: ")
year_selection = alt.selection_point(
    fields=["year"],
    bind=year_dropdown,
    value={"year": years[-1]}   # start with latest year
)

# Base: state_year joined with capital lat/lon
base = alt.Chart(state_year, title="NSF Grants per State and Year"
).transform_lookup(
    lookup="state_name",    # column in state_year (e.g. "CA", "TX", ...)
    from_=alt.LookupData(
        capitals,
        key="state",        # column in capitals dataset
        fields=["state", "lat", "lon"]
    )
).transform_filter(
    alt.datum.lat != None
).add_params(
    year_selection
).transform_filter(
    year_selection
)

# --- bubble layer ---
bubbles = base.mark_circle(
    opacity=0.9,
    stroke="white",
    strokeWidth=0.7
).encode(
    x=alt.X("lon:Q", axis=None),
    y=alt.Y("lat:Q", axis=None),
    size=alt.Size(
        "n_grants:Q",
        title="# Grants",
        scale=alt.Scale(range=[100, 4000]),  # adjust to reduce overlap
        legend=None
    ),
    color=alt.Color(
        "n_grants:Q",
        title="Number of grants",
        scale=alt.Scale(range=["#F1F4F7", "#B5D5DB", "#78B8C4", "#2F7F94"])
    ),
    tooltip=[
        alt.Tooltip("state_name:N", title="State"),
        alt.Tooltip("year:O", title="Year"),
        alt.Tooltip("n_grants:Q", title="# Grants"),
        alt.Tooltip("total_amount:Q", title="Total amount", format=",.0f")
    ]
)

# --- text labels with state abbreviation ---
labels = base.mark_text(
    fontSize=11,
    fontWeight="bold",
    dy=3,           # small vertical shift to center text
    color="white"
).encode(
    x=alt.X("lon:Q", axis=None),
    y=alt.Y("lat:Q", axis=None),
    text="state:N"   # abbreviation from capitals (CA, TX, ...)
)

chart_q1 = (bubbles + labels).properties(
    width=800,
    height=500
).configure_view(
    stroke=None
)

chart_q1
```

```python
import pandas as pd

dir_year = (
    df_clean
    .groupby(["dir_abbr", "org_dir_long_name", "year"], as_index=False)
    .agg(
        n_grants=("awd_id", "nunique"),
        total_amount=("awd_amount", "sum")
    )
)
```

```python
import altair as alt

# Unique list of years for dropdown
years = sorted(dir_year["year"].dropna().unique().tolist())

# Year selection (Altair 5 style)
year_dropdown = alt.binding_select(options=years, name="Year: ")
year_sel = alt.selection_point(fields=["year"], bind=year_dropdown)

chart_q2 = (
    alt.Chart(dir_year, title="NSF Grants per Directorate")
    .add_params(year_sel)
    .transform_filter(year_sel)
    .mark_bar()
    .encode(
        x=alt.X("n_grants:Q", title="# Grants"),
        y=alt.Y(
            "org_dir_long_name:N",
            sort="-x",
            title="Directorate"
        )
      
        ,
        tooltip=[
            alt.Tooltip("org_dir_long_name:N", title="Directorate"),
            alt.Tooltip("dir_abbr:N", title="Abbr"),
            alt.Tooltip("year:O", title="Year"),
            alt.Tooltip("n_grants:Q", title="# Grants"),
            alt.Tooltip("total_amount:Q", title="Total amount", format=",.0f"),
        ]
    )
    .properties(width=450, height=300)
)

chart_q2
```

```python
import altair as alt

# Selection that chooses the nearest year under the mouse
nearest = alt.selection_point(
    nearest=True,
    on='mouseover',
    fields=['year'],
    empty=False
)

# Base chart encoding (we'll reuse this in multiple layers)
base = alt.Chart(dir_year).encode(
    x=alt.X('year:O', title='Year'),
    y=alt.Y('n_grants:Q', title='# Grants'),
    color=alt.Color('org_dir_long_name:N', title='Directorate')
)

# 1) The main line for each directorate
line = base.mark_line().properties(
    width=800,   # wider plot like you asked
    height=400
)

# 2) Invisible selectors across the chart: capture the x (year) under the mouse
selectors = alt.Chart(dir_year).mark_point().encode(
    x='year:O',
    opacity=alt.value(0),
).add_params(
    nearest
)

# 3) Points on the lines, highlighted when nearest
points = base.mark_point().encode(
    opacity=alt.condition(nearest, alt.value(1), alt.value(0))
)

# 4) Text labels near the nearest points (show the total amount)
text = base.mark_text(align='left', dx=5, dy=-5).encode(
    text=alt.condition(nearest, 'n_grants:Q', alt.value(' '))
)

# 5) Vertical rule at the selected year
rules = alt.Chart(dir_year).mark_rule(color='gray').encode(
    x='year:O',
).transform_filter(
    nearest
)

chart_q2_line = alt.layer(
    line, selectors, points, rules, text
)

chart_q2_line
```

```python
import pandas as pd

year_summary = (
    df_clean
    .groupby("year", as_index=False)
    .agg(
        total_amount=("awd_amount", "sum"),
        n_grants=("awd_id", "nunique")
    )
)
```

```python
import altair as alt

# Selection that picks the nearest year under the mouse
nearest = alt.selection_point(
    nearest=True,
    on="mouseover",
    fields=["year"],
    empty=False
)

base = alt.Chart(year_summary).encode(
    x=alt.X("year:O", title="Year"),
    y=alt.Y("n_grants:Q", title="Number of Grants"),
)

# 1) Line of total funding
line_q4 = base.mark_line().properties(
    width=500,
    height=100,
    title="Total NSF Grant Amount Over the Years"
)

# 2) Invisible selectors across x to capture hover position
selectors_q4 = alt.Chart(year_summary).mark_point().encode(
    x="year:O",
    opacity=alt.value(0),
).add_params(
    nearest
)

# 3) Points on the line highlighted when nearest
points_q4 = base.mark_point().encode(
    opacity=alt.condition(nearest, alt.value(1), alt.value(0))
)

# 4) Text labels showing the amount for the selected year
text_q4 = base.mark_text(align="left", dx=5, dy=-5).encode(
    text=alt.condition(nearest, "n_grants:Q", alt.value(" "))
)

# 5) Vertical rule at the selected year
rules_q4 = alt.Chart(year_summary).mark_rule(color="gray").encode(
    x="year:O",
).transform_filter(
    nearest
)

chart_q4 = alt.layer(
    line_q4, selectors_q4, points_q4, rules_q4, text_q4
)
chart_q4
```
